```python
C1 = np.array([[0, 1, 2, 3],
              [1, 0, 1, 2],
              [2, 1, 0, 1],
              [3, 2, 1, 0]], dtype=float)
a1 = np.array([0.25, 0.15, 0.15, 0.45], dtype=float)
b1 = np.array([0.3, 0.15, 0.35, 0.2], dtype=float)

P1, hist_marg_1 = sinkhorn_marg_violation(a1,b1,C1,eps=0.1, prec=10**-6)
plot_hist_marg(hist_marg_1)
```

```python
P2, hist_marg_2 = sinkhorn_marg_violation(a1,b1,C1,eps=10, prec=10**-6)
P3, hist_marg_3 = sinkhorn_marg_violation(a1,b1,C1,eps=10**-2, prec=10**-6)
compare_hist_marg([hist_marg_2[2],hist_marg_1[2], hist_marg_3[2]], [10,0.1,10**-3])
```
